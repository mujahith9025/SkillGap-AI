This notebook performs structured exploratory analysis on the benchmark datasets:

```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Add src to Python path
src_path = str(Path("..").resolve() / "src")
if src_path not in sys.path:
    sys.path.append(src_path)

from data_loader import DataLoader

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

loader = DataLoader(data_dir=Path("../data"))
print("DataLoader initialized successfully.")
```

### 1. Dataset Shapes and Summary Overview

```python
for name, df in loader.datasets.items():
    print(f"{name:20}: {df.shape[0]} rows, {df.shape[1]} columns | Missing: {df.isnull().sum().sum()}")
    display(df.head(2))
```

### 2. Most In-Demand Skills Across All Roles

```python
merged_cs = pd.merge(loader.career_skills, loader.skills, on="skill_id")
skill_counts = merged_cs["skill_name"].value_counts().head(10)

plt.figure(figsize=(10, 5))
sns.barplot(x=skill_counts.values, y=skill_counts.index, palette="crest_r")
plt.title("Top 10 Most In-Demand Skills Across All Career Paths", fontsize=14, fontweight="bold")
plt.xlabel("Count of Roles Requiring Skill")
plt.ylabel("Skill Name")
plt.show()
```

### 3. Career Role vs. Skill Requirement Matrix (Heatmap)

```python
matrix = loader.get_career_skill_matrix()
plt.figure(figsize=(15, 6))
sns.heatmap(matrix, cmap="YlGnBu", linewidths=0.5, annot=False)
plt.title("Career Role vs. Skill Requirement Matrix (Importance Weights)", fontsize=14, fontweight="bold")
plt.xlabel("Skills")
plt.ylabel("Careers")
plt.xticks(rotation=45, ha="right")
plt.show()
```

### 4. Skill Importance Distribution (Core vs. Secondary vs. Optional)

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=loader.career_skills, x="importance_level", order=["Core", "Secondary", "Optional"], palette="Set2")
plt.title("Distribution of Skill Criticality (Importance Level)", fontsize=13, fontweight="bold")
plt.xlabel("Importance Level")
plt.ylabel("Count")
plt.show()
```
